```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

model_name = "ricdomolm/lawma-8b" #"meta-llama/Llama-3.2-1B-Instruct" #"dice-research/lola_v1"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name, device_map="auto")

def testar_contexto(max_tokens):
    texto = "Palavra " * max_tokens  # texto simples, repetido
    inputs = tokenizer(texto, return_tensors="pt").to(model.device)
    
    try:
        outputs = model.generate(**inputs, max_new_tokens=5)
        print(f"✅ {max_tokens} tokens: aceito!")
    except Exception as e:
        print(f"❌ {max_tokens} tokens: falhou - {e}")

# Teste incremental

for tamanho in [512, 1024, 2048, 4096, 8192, 16384, 32768, 65536, 131072, 262144]:
    testar_contexto(tamanho)
```
